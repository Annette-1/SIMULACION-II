$$

**Ejemplo3**: Calificación de $n$ estudios

Definimos
$$
X_i
\begin{cases}
1 \quad si \quad el \quad estudiante \quad i \quad aprobó \\
0 \quad otro \quad caso
\end{cases}
$$

Entonces

$$
f(x_i | \vec{θ})=θ^{x_i} \left( 1-θ \right) ^{1-x_i}
$$

$$
f(\vec{x} | \vec{θ})= Π_{i=1}^n \{ θ^{x_i} \left( 1-θ \right)^{1-x_i} \}
$$


***DEFINICIÓN:*** La familia de modelos especificados $f(\vec{x} | \vec{θ}) \quad ,\quad \vec{θ}∈θ$ se llama **familia exponencial** si $f(\vec{x} | \vec{θ})$ tiene la representación

$$
f(\vec{x} | \vec{θ})= exp \{ c(\vec{\theta}) + ∑_{j=i} ^P t_j (\vec{x})A_j (\vec{\theta}) \}h(\vec{x})
$$

Donde $c(*)$ , $A(*)$ depende solo de $\vec{θ}$ y $t(*)$ depende solo de $\vec{x}$

*Nota: $x={x_i}, i= 0,..,15$ y $\theta: (α,\beta)$*


 ---
Queremos escribir:

$$
Π(\theta | x)=Π(θ)f(x | θ) \\
Π(\theta | x)=∫_{θ'}Π(θ')f(x | θ') dθ' \quad (evidencia \quad constante)
$$

Donde:
* $Π(\theta | x)$ es anterior

* $Π(θ)$ es prior

* $f(x | θ)$ es likelihood esto ya lo hemos hecho
